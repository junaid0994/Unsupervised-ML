```python
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
import pandas as pd 
```

```python
iris = load_iris()
```

```python
X = iris.data
y = iris.target
```

```python
#visualize
sns.scatterplot(x = X[:,0] , y = X[:,2], c = y)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_scaled = sc.fit_transform(X)
```

```python
# optional - dimensionality reduciton using PCA
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

pca_data = pca.fit_transform(X_scaled)
```

```python
# Elbow method
wcss = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k)
    kmeans.fit(pca_data)
    wcss.append(kmeans.inertia_)
```

```python
sns.lineplot(x=range(1, 11), y=wcss, marker="o")
```

```python
kmeans = KMeans(
    n_clusters = 3,     
)
labels = kmeans.fit_predict(X_scaled)

```

```python
sns.scatterplot(x = X[:,0], y = X[:,2], c = labels)
```

```python
# KMeans

kmeans = KMeans(n_clusters=3, random_state=10)
labels = kmeans.fit_predict(pca_data)

sns.scatterplot(x=pca_data[:, 0], y=pca_data[:,1], c=labels)
sns.scatterplot(x=kmeans.cluster_centers_[:, 0], y=kmeans.cluster_centers_[:, 1], marker="X", c="red", s=200)
```
